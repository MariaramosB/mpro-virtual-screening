# 02 — Structure Preparation

## Objective

Prepare the SARS-CoV-2 main protease (Mpro) structure for subsequent molecular docking.

## Input structure

- **PDB:** 7BQY
- **Protein:** SARS-CoV-2 Mpro
- **Chain:** A
- **Length:** 306 aa
- **Resolution:** 1.70 Å
- **Experimental method:** X-ray crystallography

## Reference ligand

The crystal structure contains the covalent inhibitor N3.

N3 is represented by multiple chemical components in the PDB and is covalently linked to Cys145.

N3 will be used as a structural reference during protocol development and validation, rather than as a conventional non-covalent docking ligand.

## Catalytic dyad

- His41
- Cys145

## Workflow

1. Inspect the original PDB structure.
2. Identify protein and non-protein components.
3. Define the receptor structure.
4. Prepare the receptor for docking.
5. Verify the prepared structure.
6. Save the processed receptor for subsequent docking steps.
